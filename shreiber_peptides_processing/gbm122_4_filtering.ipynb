# 4. Apply filtering

```python
def filter_epitopes(df, method_name, advantage = False):
    if method_name == 'netMHC-3.4':
        df = df.loc[df[method_name + "_affinity_mt"] < 500].copy()
        if advantage:
            df = df.loc[df[method_name + "_score_advantage"] > 1].copy()
    elif  method_name == 'netMHC-4.0':
        df = df.loc[df[method_name +  "_percentile_rank_mt"] < 2].copy()
        if advantage:
            df = df.loc[df[method_name + "_percentile_rank_advantage"] > 1].copy()
    elif method_name == 'netMHCpan-4.1':
        df = df.loc[df[method_name + "_percentile_rank_el_mt"] < 2].copy()
        if advantage:
            df = df.loc[df[method_name + "_percentile_rank_el_advantage"] > 1].copy()
    elif method_name == 'mhcnuggets':
        df = df.loc[df[method_name + "_percentile_rank_mt"] < 2].copy()
        if advantage:
            df = df.loc[df[method_name + "_percentile_rank_advantage"] > 1].copy()
    elif method_name == 'acme':
        df = df.loc[df[method_name + "_score_mt"] > 0.42].copy()
        if advantage:
            df = df.loc[df[method_name + "_score_advantage"] > 1].copy()
    elif method_name == 'mhcseqnet':
        df = df.loc[df[method_name + "_score_mt"] > 0.99].copy()
        if advantage:
            df = df.loc[df[method_name + "_score_advantage"] > 1].copy()
    elif method_name == 'puffin':
        df = df.loc[df[method_name + "_score_mt"] > 0.42].copy()
        if advantage:
            df = df.loc[df[method_name + "_score_advantage"] > 1].copy()
#     elif method_name == 'puffin':
#         df = df.loc[df[method_name + "_binding_likelihood_mt"] > 0.5].copy()
#         if advantage:
#             df = df.loc[df[method_name + "_binding_likelihood_advantage"] > 1].copy()
    elif method_name == 'mhcattnnet':
        df = df.loc[df[method_name + "_score_mt"] >= 1].copy()
        if advantage:
            df = df.loc[df[method_name + "_score_advantage"] > 1].copy()
    return df
```

```python
unsupported_alleles = { "acme": {"HLA-C*04:01", "HLA-C*03:04"},
                        "netMHCpan-4.1" : set(),
                        "puffin" : {"HLA-C*04:01"},
                        "mhcnuggets" : set(),
                        "mhcseqnet" : set(),
                        "netMHC-4.0" : {"HLA-C*03:04"},
                        "netMHC-3.4" : {"HLA-C*03:04"},
                         "mhcattnnet": set()                
}
```

```python
def per_allele_filter_predictions(per_allele_df, selected_methods, advantage=False):
    per_allele_result = dict()
    for allele_name in ALLELES:
        prediction_df =  per_allele_df[allele_name]
        count = len(prediction_df)
        print(f"{allele_name}:({count}) ->", end=" ")
        for method in selected_methods:
            if allele_name not in unsupported_alleles[method]:
                prediction_df = filter_epitopes(prediction_df, method, advantage)
                count = len(prediction_df)
                print(f"{method}:({count}) ->", end=" ")
        print()
        per_allele_result[allele_name] = prediction_df
    return per_allele_result

def per_allele_add_source_info(per_allele_df, region):
    per_allele_result = dict()
    df_source =  per_region_total_df[region].copy()
    df_source = df_source.rename(columns={ "HLA Allele":"allele", "MT Epitope Seq":"peptide_mt", "WT Epitope Seq":"peptide_wt"}) # rename for convinience
    
    # No Transcript, Protein position , Transcript Expression, Transcript Support Level because it produces many duplicates
    df_source = df_source[
        [ 'allele','peptide_mt', 'peptide_wt','Chromosome', 'Start', 'Stop', 'Reference', 'Variant', 
         'Ensembl Gene ID', 
         'Variant Type','Mutation', 
         'Gene Name', 'HGVSc', 'HGVSp', 'Mutation Position', 
         'Tumor DNA Depth', 'Tumor DNA VAF', 'Tumor RNA Depth', 'Tumor RNA VAF',
         'Normal Depth', 'Normal VAF', 'Gene Expression',
         'cterm_7mer_gravy_score','max_7mer_gravy_score', 'difficult_n_terminal_residue',
       'c_terminal_cysteine', 'c_terminal_proline', 'cysteine_count','n_terminal_asparagine', 'asparagine_proline_bond_count']].drop_duplicates()
    
    
    for allele_name in ALLELES:
        df_top = per_allele_df[allele_name].copy()
        len_initial = len(df_top)
        df_merged = df_source.merge(df_top, on=["allele","peptide_mt", "peptide_wt" ])
        len_merged = len(df_merged)
        df_filtered = df_source[(df_source['peptide_mt'].isin(df_top['peptide_mt'])) & (df_source['peptide_wt'].isin(df_top['peptide_wt'])) & (df_source['allele'].isin(df_top['allele']))]
        len_filtered = len(df_filtered)
        print(f"top epitopes {len_initial}, after merging  {len_merged}, after filtering {len_filtered}")
        if not len_merged == len_initial :
            print(f"many rows for 1 epitope due to multiple Gene ID")
        
        per_allele_result[allele_name] = df_merged
    return per_allele_result

def filter_using_aux_data(df):
    df_filtered = df.copy()
    # VAF is the percentage of sequence reads observed matching a specific DNA variant divided by the overall coverage at that locus
    df_filtered = df_filtered[df_filtered['Tumor DNA VAF'] > 0.1]
    df_filtered = df_filtered[df_filtered['Tumor RNA VAF'] > 0.1]
    # Coverage (or depth) in DNA sequencing is the number of unique reads that include a given nucleotide in the reconstructed sequence
    df_filtered = df_filtered[df_filtered['Tumor DNA Depth'] > 10]
    df_filtered = df_filtered[df_filtered['Tumor RNA Depth'] > 10]
    
    # Coverage (or depth) in DNA sequencing is the number of unique reads that include a given nucleotide in the reconstructed sequence
    df_filtered = df_filtered[df_filtered['Gene Expression'] > 1]
    return df_filtered

def reorder_result_columns(df_filtered):  
    df_filtered = df_filtered[
        [ 
            #base info
            'region', 'allele', 'peptide_mt', 'peptide_wt', 
            #scores
           'netMHC-3.4_score_mt', 'netMHC-3.4_affinity_mt', 'netMHC-3.4_percentile_rank_mt',
           'netMHC-4.0_score_mt', 'netMHC-4.0_affinity_mt',
           'netMHC-4.0_percentile_rank_mt', 'netMHCpan-4.1_score_ba_mt',
           'netMHCpan-4.1_score_el_mt', 'netMHCpan-4.1_affinity_mt',
           'netMHCpan-4.1_percentile_rank_ba_mt',
           'netMHCpan-4.1_percentile_rank_el_mt', 'puffin_score_mt',
           'puffin_epistemic_var_mt', 'puffin_aleatoric_var_mt',
           'puffin_binding_likelihood_mt', 'mhcnuggets_score_mt',
           'mhcnuggets_affinity_mt', 'mhcnuggets_percentile_rank_mt',
           'acme_score_mt', 'mhcseqnet_score_mt', 'mhcattnnet_score_mt',
           'netMHC-3.4_score_advantage', 'netMHC-3.4_affinity_advantage',
           'netMHC-4.0_score_advantage', 'netMHC-4.0_affinity_advantage',
           'netMHC-4.0_percentile_rank_advantage',
           'netMHCpan-4.1_score_ba_advantage', 'netMHCpan-4.1_score_el_advantage',
           'netMHCpan-4.1_affinity_advantage',
           'netMHCpan-4.1_percentile_rank_ba_advantage',
           'netMHCpan-4.1_percentile_rank_el_advantage', 'puffin_score_advantage',
           'puffin_epistemic_var_advantage', 'puffin_aleatoric_var_advantage',
           'puffin_binding_likelihood_advantage', 'mhcnuggets_score_advantage',
           'mhcnuggets_affinity_advantage', 'mhcnuggets_percentile_rank_advantage',
           'acme_score_advantage', 'mhcseqnet_score_advantage',
           'mhcattnnet_score_advantage', 'netMHC-3.4_score_wt',
           'netMHC-3.4_affinity_wt', 'netMHC-3.4_percentile_rank_wt',
           'netMHC-4.0_score_wt', 'netMHC-4.0_affinity_wt',
           'netMHC-4.0_percentile_rank_wt', 'netMHCpan-4.1_score_ba_wt',
           'netMHCpan-4.1_score_el_wt', 'netMHCpan-4.1_affinity_wt',
           'netMHCpan-4.1_percentile_rank_ba_wt',
           'netMHCpan-4.1_percentile_rank_el_wt', 'puffin_score_wt',
           'puffin_epistemic_var_wt', 'puffin_aleatoric_var_wt',
           'puffin_binding_likelihood_wt', 'mhcnuggets_score_wt',
           'mhcnuggets_affinity_wt', 'mhcnuggets_percentile_rank_wt',
           'acme_score_wt', 'mhcseqnet_score_wt', 'mhcattnnet_score_wt', 
            # source info
             'Chromosome', 'Start', 'Stop',
           'Reference', 'Variant',
           'Ensembl Gene ID', 'Variant Type', 'Mutation',
           'Gene Name', 'HGVSc', 'HGVSp', 'Mutation Position', 'Tumor DNA Depth',
           'Tumor DNA VAF', 'Tumor RNA Depth', 'Tumor RNA VAF', 'Normal Depth',
           'Normal VAF', 'Gene Expression',
           'cterm_7mer_gravy_score', 'max_7mer_gravy_score',
           'difficult_n_terminal_residue', 'c_terminal_cysteine',
           'c_terminal_proline', 'cysteine_count', 'n_terminal_asparagine',
           'asparagine_proline_bond_count'
        ]
    ]
    df_filtered = df_filtered.drop(columns=['netMHC-3.4_percentile_rank_mt', 'netMHC-3.4_percentile_rank_wt'])

    return df_filtered


def filter_and_save_to_file(per_region_per_allele_predictions, selected_methods, advantage=False):
    print("1.Filter using methods")
    per_region_per_allele_top_epitopes = dict()
    for region in REGIONS:
        per_region_per_allele_top_epitopes[region] = dict()
        print(f"region {region}")
        per_region_per_allele_top_epitopes[region] = per_allele_filter_predictions(per_region_per_allele_predictions[region], selected_methods, advantage=advantage)
    print("2.Add aux info")
    per_region_per_allele_top_epitopes_with_info = dict()
    for region in REGIONS:
        per_region_per_allele_top_epitopes_with_info[region] = per_allele_add_source_info(per_region_per_allele_top_epitopes[region], region)
    print("3.Concat result dfs")
    list_of_dfs = list()
    for region in REGIONS:
        for allele_name in ALLELES:
            df = per_region_per_allele_top_epitopes_with_info[region][allele_name]
            df['region'] = f"Region_{region}"
            list_of_dfs.append(df)
        
    result_df = pd.concat(list_of_dfs)
    print("4.Filter using thesholds for aux data")
    df_filtered = filter_using_aux_data(result_df)
    print("5.Reorder columns")
    df_filtered = reorder_result_columns(df_filtered)
    print("6.Write to file")
    #DATA_PATH = "/home/denklewer/Data/gbm122/" #ubuntu machine
    DATA_PATH = "D:/Google Drive/Work/ITMO/MHC-Epitop-Prediction/prediction_results/gbm122/" # windows
    if advantage:
        df_filtered.to_csv("{}top_epitopes_filtered+advantage_all_regions_{}.csv".format(DATA_PATH, "_".join(selected_methods)),sep=";", index=False)
    else:
        df_filtered.to_csv("{}top_epitopes_filtered_all_regions_{}.csv".format(DATA_PATH, "_".join(selected_methods)),sep=";", index=False)
    return df_filtered
```

```python
selected_methods = ['netMHCpan-4.1','mhcnuggets', 'acme','puffin','mhcseqnet',]
df_filtered = filter_and_save_to_file(per_region_per_allele_predictions, selected_methods, advantage=False)
```
