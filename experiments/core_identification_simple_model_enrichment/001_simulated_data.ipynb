## Length distributions

```python
from pomegranate import *
### Length distributions
center_dist_examples = [
    {
    12: 357,
    13: 812,
    14: 1251,
    15: 5369,
    16: 1449,
    17: 991,
    18: 661
},
    {
    12: 148,
    13: 458,
    14: 656,
    15: 3117,
    16: 900,
    17: 585,
    18:327
}
]

length_distributions = list()
## 15 mers is preferrable
for i in range(len(motifs)):
    # generate random length distribution based on one example (random choice after motif num > 1)
    d_length_motif = DiscreteDistribution()
    if i >= 2:
        source_dist = np.random.choice(center_dist_examples)
    else:
        source_dist = center_dist_examples[i]
    fit_data = list()
    for target_length, example_cnt in source_dist.items():
        fit_data += [target_length]* (example_cnt + np.random.randint(-example_cnt*0.2, example_cnt*0.2))
    d_length_motif.fit(fit_data)
    length_distributions.append(d_length_motif)

```

## Start distributions

```python
import scipy.stats as ss
## Start distributions

## in the middle is more possible
def get_start_dist_middle(target_length= 15, motif_len=9):
    x = np.arange(0, target_length - (motif_len-1))
    xU, xL = x + 0.5, x - 0.5
    mean_x = np.mean(x)
    probs = ss.norm.cdf(xU, np.floor(mean_x), scale = 1.8) - ss.norm.cdf(xL, np.floor(mean_x),  scale = 1.8)
    probs = probs / probs.sum() # normalize the probabilities so their sum is 1
    d_dict = {key: prob for key, prob in zip(x,probs)}
    dist = DiscreteDistribution(d_dict)
    return dist

core_lengths = [max(motif.keys()) + 1 for motif in motifs]

start_distributions = [
    {
    target_length : get_start_dist_middle(target_length, core_lengths[i]) for target_length in LENGTHS_TO_GENERATE
    }
    for i in range(len(motifs))]
```

```python

from collections import defaultdict

def get_new_distribution(motifs=motifs, motif_num=-1, position=0):
    d_dict = {key: prob for key, prob in zip(amino_acids_list, np.random.default_rng().uniform(0, 1, size=len(amino_acids_list)))}
    if (motif_num) >= 0:
        anchors = motifs[motif_num]
        if position in anchors:
            d_dict = {key: 0 for key in amino_acids_list}
            for amino_acid in anchors[position]:
                base_value = np.random.default_rng().uniform(0, 1)
                d_dict[amino_acid] = base_value
    d_dict = {key: prob/sum(d_dict.values()) for key, prob in d_dict.items()}
    return DiscreteDistribution(d_dict)


```

```python
def generate_cores(motifs, n= 5000, ratio_combination=RATIOS_TO_GENERATE[0], noize_level=NOIZE_LEVELS_TO_GENERATE[0]):
    motifs_selection_distribution = DiscreteDistribution({i:ratio for i, ratio in enumerate(ratio_combination)})
    noize_selection_distribution = DiscreteDistribution({1:noize_level, 0:1-noize_level})

    motifs_aa_distributions = [
        {
            pos: get_new_distribution(motifs, motif_num=i, position=pos) for pos in range(core_lengths[i])
        }
        for i in range(len(motifs))
    ]
    cores = np.empty(shape=(n),dtype=object)
    #cores = np.empty((n, core_len), dtype=str)
    peptide_labels = motifs_selection_distribution.sample(n)
    noize_labels = noize_selection_distribution.sample(n)

    for current_motif, motif in enumerate(motifs_aa_distributions):
        core_len = core_lengths[current_motif]
        idx_values = (peptide_labels == current_motif)
        curr_motif_cnt = len(peptide_labels[idx_values])
        curr_noize_cnt = len(peptide_labels[(noize_labels == 1) & (peptide_labels == current_motif)])
        print(f"For motif {current_motif} generate {curr_motif_cnt} values, but {curr_noize_cnt} will be replaced by noize")
        current_cores =  np.empty((curr_motif_cnt, core_len), dtype=str)
        for pos in range(core_len):
            current_cores[:, pos] = motif[pos].sample(curr_motif_cnt)
        print(f"shape of cores {current_cores.shape}")
        print(f"example of core {''.join(current_cores[0,])}")
        list_of_cores = list(np.apply_along_axis(lambda row: "".join(row), axis=1, arr=current_cores))
        cores[idx_values] = list_of_cores
        current_noizy_cores =  np.empty((curr_noize_cnt, core_len), dtype=str)
        idx_values = (noize_labels == 1) & (peptide_labels == current_motif)
        for pos in range(core_len):
            noize_dist = get_new_distribution()
            current_noizy_cores[:, pos] = noize_dist.sample(curr_noize_cnt)
        print(f"shape of noized cores {current_noizy_cores.shape}")
        if len(current_noizy_cores) > 0:
            list_of_cores = list(np.apply_along_axis(lambda row: "".join(row), axis=1, arr=current_noizy_cores))
            cores[idx_values] = list_of_cores
        #cores.extend(current_cores)
    return cores, peptide_labels, noize_labels



def generate_peptides(motifs, cores, peptide_labels, noize_labels):
    # fill lengths
    peptide_lengths = np.zeros(len(cores))
    for current_motif, motif in enumerate(motifs):
        length_dist = length_distributions[current_motif]
        curr_motif_cnt =  len(peptide_labels[peptide_labels == current_motif])
        peptide_lengths[peptide_labels == current_motif] = length_dist.sample(curr_motif_cnt)

    # fill core starts
    core_starts = np.zeros(len(cores))
    for curr_motif, motif in enumerate(motifs):
        per_length_start_dist = start_distributions[curr_motif]
        for target_length in LENGTHS_TO_GENERATE:
            idx_values = (peptide_lengths == target_length) & (peptide_labels == curr_motif)
            cnt_for_subset = len(peptide_labels[idx_values])

            core_starts[idx_values] = per_length_start_dist[target_length].sample(cnt_for_subset)
            core_len = core_lengths[curr_motif]
            if target_length == 14:
                print(sum( core_starts[idx_values] > 4))
    # fill rest amino amino acids (flanks)
    flank_dist = get_new_distribution()
    result_cores = cores
    result_lengths = peptide_lengths.astype(int)
    result_core_starts = core_starts.astype(int)
    result_peptides = list()
    result_alleles = np.array([f"Dummy_allele_{curr_motif}" for curr_motif in peptide_labels])

    for  core_start,core, peptide_length in zip(result_core_starts, result_cores, result_lengths):
        # print(core_start,  peptide_length)
        core_len = len(core)
        left_flank = "".join(flank_dist.sample(core_start))
        right_flank_length = peptide_length - core_len - len(left_flank)
        right_flank = "".join(flank_dist.sample(right_flank_length))
        pep = left_flank + core + right_flank
        assert( len(pep) == peptide_length), (core_start, len(left_flank),  len(right_flank), peptide_length)
        result_peptides.append(pep)
       # break
    df = pd.DataFrame({
        "peptide": result_peptides,
        "core":result_cores,
        "core_start":result_core_starts,
        "length" :result_lengths,
        "allele":result_alleles,
        "noize": noize_labels,
        "binder": 1
    })

    return  df


```

```python
from pathlib import Path
Path(RESULT_DATA_PATH).mkdir(parents=True, exist_ok=True)
lengths_generated = tuple(sorted(set(max(motif.keys()) for motif in motifs)))

results = dict()
for noize_level in NOIZE_LEVELS_TO_GENERATE:
    for ratio_combination in sum_to_one_combinations:
        cores, labels, noize_labels= generate_cores(motifs, n=12000, ratio_combination=ratio_combination,  noize_level=noize_level)
        df = generate_peptides(motifs, cores, labels, noize_labels)
        df.to_csv(f"{RESULT_DATA_PATH}/sim_data_motifs_{len(motifs)}_ratio_{ratio_combination}_lengths_{lengths_generated}_noize_{noize_level}.csv",sep=";", index=False)
        results[ratio_combination] = df
```

```python
import seaborn as sns
import logomaker
def make_logo_for_data(binders_array, name, ax=None):
    sns.set_theme(style="white")
    matrix = logomaker.alignment_to_matrix(binders_array)
    info_matrix = logomaker.transform_matrix(matrix, from_type='counts', to_type='information')
    prob_matrix = logomaker.transform_matrix(matrix, from_type='counts', to_type='probability')
    ww_logo = logomaker.Logo(info_matrix, font_name='Arial Rounded MT Bold', flip_below=True, ax=ax)
    ww_logo.ax.set_title(f"{name}, {len(binders_array)}")
```

```python
def make_logo_restricted(df, peptide_len, core_start=-1, core_len = 9):
    target_df = df.copy()
    target_df = target_df[target_df.length == peptide_len]
    name = f"peptide_len: {peptide_len}"
    if core_start >= 0:
        target_df = target_df[target_df.core_start == core_start]
        name = name + f" core_start: {core_start}"
    make_logo_for_data(target_df.peptide, name)
```
