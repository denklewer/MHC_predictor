```python
EXPERIMENT_NAME = 'core_identification'
DATA_PATH_BASE = 'G:/My Drive/Work/ITMO/data'
DATA_PATH = f'{DATA_PATH_BASE}/genentech' # work laptop
FILE_NAME = '230112-classii-peptides-w-submotifs.csv'
TARGET_PATH_TO_FILES = f"{DATA_PATH_BASE}/{EXPERIMENT_NAME}"
```

```python
import pandas as pd
import sys
import os
```

```python
PATH_TO_PREDICTOR_HOME = "../.."
sys.path.append(PATH_TO_PREDICTOR_HOME)
```

```python
total_df = pd.read_csv(f"{DATA_PATH}/{FILE_NAME}")
total_df['binder'] = 1
total_df['allele'] = total_df.allele.str.replace('___', '-')
total_df['length'] = total_df.peptide.str.len()
ALLELES = total_df.allele.unique()
```

```python
per_allele_df = {allele_name: total_df[total_df.allele == allele_name].copy() for allele_name in ALLELES}
```

```python
per_allele_df[ALLELES[0]]
```

```python
import mhcnames
import os
if not os.path.exists(TARGET_PATH_TO_FILES ):
    os.makedirs(TARGET_PATH_TO_FILES )
```

```python
compact_name = mhcnames.compact_allele_name(ALLELES[0].replace('___', '-'))
for allele_name in ALLELES:
    total_df = per_allele_df[allele_name]
    compact_name = mhcnames.compact_allele_name(allele_name)
    # allele_name = allele_name.replace('-', '_').replace('*', '_').replace(':', '_')
    total_df.to_csv("{}/data_{}.csv".format(TARGET_PATH_TO_FILES,compact_name),sep=";", index=False)
```
