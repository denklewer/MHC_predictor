```python
import itertools
per_allele_per_kfold_per_length_binders_train_new = dict()
per_allele_per_kfold_per_length_binders_test_new = dict()
NEW_ALELLES = list()
selected_lengths = np.arange(8,20)
for allele_name1, allele_name2 in itertools.combinations(ALLELES, 2):
    allele_name = f"{allele_name1}+{allele_name2}"
    NEW_ALELLES.append(allele_name)
    print(allele_name)
    per_allele_per_kfold_per_length_binders_train_new[allele_name] = dict()
    per_allele_per_kfold_per_length_binders_test_new[allele_name] = dict()

    for split_num in per_allele_per_kfold_per_length_binders_train[allele_name1].keys():
        per_allele_per_kfold_per_length_binders_train_new[allele_name][split_num] = dict()
        per_allele_per_kfold_per_length_binders_test_new[allele_name][split_num] = dict()
        lengths1 = set(per_allele_per_kfold_per_length_binders_train[allele_name1][split_num].keys())
        lengths2 = set(per_allele_per_kfold_per_length_binders_train[allele_name2][split_num].keys())
        intersected_lengths = lengths1.intersection(lengths2)

        for target_length in intersected_lengths:
            data_array1 = per_allele_per_kfold_per_length_binders_train[allele_name1][split_num][target_length]
            data_array2 = per_allele_per_kfold_per_length_binders_train[allele_name2][split_num][target_length]
            per_allele_per_kfold_per_length_binders_train_new[allele_name][split_num][target_length] = np.concatenate([data_array1, data_array2])
            if split_num == 0:
                print(f"allele {allele_name1}, length {target_length}: {len(data_array1)}")
                print(f"allele {allele_name2}, length {target_length}: {len(data_array2)}")

            data_array1 = per_allele_per_kfold_per_length_binders_test[allele_name1][split_num][target_length]
            data_array2 = per_allele_per_kfold_per_length_binders_test[allele_name2][split_num][target_length]
            per_allele_per_kfold_per_length_binders_test_new[allele_name][split_num][target_length] = np.concatenate([data_array1, data_array2])


per_allele_per_kfold_per_length_binders_train = per_allele_per_kfold_per_length_binders_train_new
per_allele_per_kfold_per_length_binders_test = per_allele_per_kfold_per_length_binders_test_new
ALLELES = NEW_ALELLES
```
