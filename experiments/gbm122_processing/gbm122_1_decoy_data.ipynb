## 0.1 Read source decoy data file

```python
DECOY_DATA_PATH = "/home/denklewer/Data/decoy/"
```

```python
def filter_canonical_aa(df):
    df = df[~df['peptide'].str.contains("U")]
    df = df[~df['peptide'].str.contains("\+")]
    df = df[~df['peptide'].str.contains("X")]
    df = df[~df['peptide'].str.contains("B")]
    df = df[~df['peptide'].str.contains("Z")]
    df = df[~df['peptide'].str.contains("O")]
    df = df[~df['peptide'].str.contains("J")]
    df = df[~df['peptide'].str.contains("-")]
    return df
```

```python
decoy_source_human_class1 = pd.read_csv("{}decoy-human-class-i-2.txt".format(DECOY_DATA_PATH), header=None, names = ["peptide"])
decoy_source_human_class1 = filter_canonical_aa(decoy_source_human_class1.drop_duplicates())
decoy_source_human_class2 = pd.read_csv("{}decoy-human-class-ii-1.txt".format(DECOY_DATA_PATH), header=None, names = ["peptide"])
decoy_source_human_class2 = filter_canonical_aa(decoy_source_human_class2.drop_duplicates())

decoy_source_mouse_class1 = pd.read_csv("{}decoy-murine-class-i-2.txt".format(DECOY_DATA_PATH), header=None, names = ["peptide"])
decoy_source_mouse_class1 = filter_canonical_aa(decoy_source_mouse_class1.drop_duplicates())
decoy_source_mouse_class2 = pd.read_csv("{}decoy-murine-class-ii-1.txt".format(DECOY_DATA_PATH), header=None, names = ["peptide"])

decoy_source_mouse_class2 = filter_canonical_aa(decoy_source_mouse_class2.drop_duplicates())
```

## 0.2 Sample peptides

```python
# sample 200000
SAMPLE_SIZE = 200_000
sampled_df_human_1 = decoy_source_human_class1.sample(SAMPLE_SIZE)
sampled_df_human_2 = decoy_source_human_class2.sample(SAMPLE_SIZE)
sampled_df_mouse_1 = decoy_source_mouse_class1.sample(SAMPLE_SIZE)
sampled_df_mouse_2 = decoy_source_mouse_class2.sample(SAMPLE_SIZE)

```

## 0.3 Save to file

```python
sampled_df_human_1.to_csv("{}sampled_decoy_data_human_class_I.csv".format(DECOY_DATA_PATH), index=False)
sampled_df_human_2.to_csv("{}sampled_decoy_data_human_class_II.csv".format(DECOY_DATA_PATH), index=False)
sampled_df_mouse_1.to_csv("{}sampled_decoy_data_mouse_class_I.csv".format(DECOY_DATA_PATH), index=False)
sampled_df_mouse_2.to_csv("{}sampled_decoy_data_mouse_class_II.csv".format(DECOY_DATA_PATH), index=False)
```
